```python
import manim as mn
from manim import *
import numpy as np

config.media_width = "75%"
config.verbosity = "WARNING"

print(mn.__version__)
```

```python
%%manim -qm IntroScene

class IntroScene(Scene):

    def construct(self):

        title = Text("Can a Robot Touch Any Point?")
        self.play(Write(title))
        self.wait()

        self.play(title.animate.to_edge(UP))

        base = Dot(ORIGIN)

        arm = Line(
            ORIGIN,
            3*RIGHT,
            stroke_width=10
        )

        target = Dot(
            2*RIGHT + 1.5*UP,
            color=RED
        )

        self.play(Create(base))
        self.play(Create(arm))
        self.play(FadeIn(target))

        question = Text(
            "How should the robot move?",
            font_size=36
        ).next_to(target, RIGHT)

        self.play(Write(question))

        self.wait(2)
```

```python
%%manim -qm OnlyMotors


class OnlyMotors(Scene):

    def construct(self):

        base = Dot()

        arm = Line(
            ORIGIN,
            3*RIGHT,
            stroke_width=10
        )

        joint = Dot()

        theta = MathTex(r"\theta")

        theta.next_to(joint, DOWN)

        self.play(Create(base))
        self.play(Create(arm))
        self.play(FadeIn(theta))

        self.play(
            Rotate(
                arm,
                angle=PI/2,
                about_point=ORIGIN
            ),
            run_time=2
        )

        self.wait()
```

```python
%%manim -qm TwoLinkPreview

class TwoLinkPreview(Scene):

    def construct(self):

        theta1 = ValueTracker(.5)
        theta2 = ValueTracker(1.2)

        base = Dot()

        arm = always_redraw(lambda:

            VGroup(

                Line(
                    ORIGIN,
                    2*np.array([
                        np.cos(theta1.get_value()),
                        np.sin(theta1.get_value()),
                        0
                    ]),
                    stroke_width=10
                ),

                Line(

                    2*np.array([
                        np.cos(theta1.get_value()),
                        np.sin(theta1.get_value()),
                        0
                    ]),

                    2*np.array([
                        np.cos(theta1.get_value()),
                        np.sin(theta1.get_value()),
                        0
                    ]) +

                    1.5*np.array([
                        np.cos(theta1.get_value()+theta2.get_value()),
                        np.sin(theta1.get_value()+theta2.get_value()),
                        0
                    ]),

                    stroke_width=10

                )

            )

        )

        self.add(base, arm)

        self.play(theta1.animate.set_value(2.5), run_time=3)
        self.play(theta2.animate.set_value(-1.5), run_time=3)

        self.wait()
```

```python
%%manim -qm RotatingStick

class RotatingStick(Scene):

    def construct(self):

        theta = ValueTracker(0)

        arm = always_redraw(
            lambda:
            Line(
                ORIGIN,
                3*np.array([
                    np.cos(theta.get_value()),
                    np.sin(theta.get_value()),
                    0
                ]),
                stroke_width=10
            )
        )

        endpoint = always_redraw(
            lambda:
            Dot(
                3*np.array([
                    np.cos(theta.get_value()),
                    np.sin(theta.get_value()),
                    0
                ]),
                color=RED
            )
        )

        trail = TracedPath(
            endpoint.get_center,
            stroke_color=YELLOW
        )

        self.add(trail, arm, endpoint)

        self.play(
            theta.animate.set_value(TAU),
            run_time=8,
            rate_func=linear
        )

        self.wait()
```

```python
%%manim -qm RightTriangle


class RightTriangle(Scene):

    def construct(self):

        angle = 40*DEGREES
        L = 3

        end = L*np.array([
            np.cos(angle),
            np.sin(angle),
            0
        ])

        arm = Line(ORIGIN,end)

        horiz = Line(ORIGIN,[end[0],0,0],color=BLUE)

        vert = Line([end[0],0,0],end,color=GREEN)

        self.play(Create(arm))
        self.play(Create(horiz))
        self.play(Create(vert))

        self.wait()
```
