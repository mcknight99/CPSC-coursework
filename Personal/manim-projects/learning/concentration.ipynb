```python
# calculating the % concentration of a batch of solution that gets partially emptied, refilled with water, adding an additive, and repeating to find the end behavior of concentration

# ex. a barrel of 55 gal water and 1/128 gal (1 fl oz) of fluid additive starts in a barrel
# then, after ideal mixing, the barrel is emptied to 2 gallons of the mixture
# the barrel is then refilled to 55 gallons using water, then adding another 1/128 gal of fluid additive. print the result of % concentration after each cycle


import math

def calculate_concentration(initial_volume, initial_additive, empty_volume, refill_volume, added_additive, cycles):
    # Calculate initial concentration
    concentration = (initial_additive / (initial_volume + initial_additive))
    print(f"Cycle 0: \t{concentration*100:.20f}%")

    for cycle in range(1, cycles + 1):
        # Calculate the amount of additive left after emptying to empty_volume
        empty_concentration = concentration * empty_volume
        
        # Refill the barrel with water and add the new additive
        total_additive = (empty_concentration * empty_volume) + added_additive
        
        # Calculate new concentration
        concentration = (total_additive / (refill_volume + added_additive))
        
        print(f"Cycle {cycle}: \t{concentration*100:.20f}%")

# Example usage
initial_volume = 55  # gallons
initial_additive = 1 / 128  # gallons (1 fl oz)
empty_volume = 2  # gallons
refill_volume = 55  # gallons
added_additive = 1 / 128  # gallons (1 fl oz)
cycles = 100  # number of cycles to simulate

calculate_concentration(initial_volume, initial_additive, empty_volume, refill_volume, added_additive, cycles)

```

```python
# use manim to graph the concentration over time, with a line graph showing the % concentration on the y-axis and the cycle number on the x-axis. The graph should show how the concentration approaches a limit as the number of cycles increases.

import manim as mn
from manim import *

config.media_width = "75%"
config.verbosity = "WARNING"

print(mn.__version__)
```

```python
%%manim -ql ConcentrationAnimation

from manim import *

class ConcentrationAnimation(Scene):

    def make_bar(self, water, additive, scale=4, width=1):
        """
        Create a stacked bar representing the current mixture.
        The total height is always 'scale'.
        """

        total = water + additive

        water_height = scale * water / total
        additive_height = scale * additive / total

        # Water rectangle
        water_rect = Rectangle(
            width=width,
            height=water_height,
            fill_color=BLUE,
            fill_opacity=1,
            stroke_color=WHITE,
        )

        # Put its bottom at y = -scale/2
        water_rect.move_to(DOWN * (scale/2 - water_height/2))

        # Additive rectangle
        additive_rect = Rectangle(
            width=width,
            height=additive_height,
            fill_color=RED,
            fill_opacity=1,
            stroke_color=WHITE,
        )

        additive_rect.next_to(water_rect, UP, buff=0)

        outline = Rectangle(
            width=width,
            height=scale,
            color=WHITE,
        )

        return VGroup(outline, water_rect, additive_rect)

    def construct(self):

        ####################################################
        # Parameters
        ####################################################

        initial_water = 55
        initial_additive = 1 / 128

        refill_water = 55
        added_additive = 1 / 128

        remaining_volume = 2

        cycles = 10

        ####################################################
        # Initial state
        ####################################################

        water = initial_water
        additive = initial_additive

        bar = self.make_bar(water, additive)

        title = Text("Cycle 0", font_size=32).to_edge(UP)

        info = Text("", font_size=24).next_to(bar, RIGHT, buff=1)

        self.add(bar, title, info)

        ####################################################
        # Animation loop
        ####################################################

        for cycle in range(cycles):

            ################################################
            # Display current concentration
            ################################################

            concentration = 100 * additive / (water + additive)

            new_info = Text(
                f"Water: {water:.3f} gal\n"
                f"Additive: {additive:.5f} gal\n"
                f"Conc: {concentration:.5f}%",
                font_size=24,
            ).next_to(bar, RIGHT, buff=1)

            self.play(Transform(info, new_info))
            self.wait(0.5)

            ################################################
            # Empty barrel to remaining_volume gallons
            ################################################

            fraction = remaining_volume / (water + additive)

            water *= fraction
            additive *= fraction

            new_bar = self.make_bar(water, additive)

            empty_label = Text(
                f"Cycle {cycle+1}: Empty to {remaining_volume} gal",
                font_size=32
            ).to_edge(UP)

            self.play(
                Transform(bar, new_bar),
                Transform(title, empty_label),
                run_time=1.5
            )

            ################################################
            # Refill with water
            ################################################

            water = refill_water

            new_bar = self.make_bar(water, additive)

            refill_label = Text(
                f"Cycle {cycle+1}: Refill with Water",
                font_size=32
            ).to_edge(UP)

            self.play(
                Transform(bar, new_bar),
                Transform(title, refill_label),
                run_time=1.5
            )

            ################################################
            # Add fresh additive
            ################################################

            additive += added_additive

            new_bar = self.make_bar(water, additive)

            add_label = Text(
                f"Cycle {cycle+1}: Add Additive",
                font_size=32
            ).to_edge(UP)

            self.play(
                Transform(bar, new_bar),
                Transform(title, add_label),
                run_time=1.5
            )

        ####################################################
        # Final concentration
        ####################################################

        concentration = 100 * additive / (water + additive)

        final_info = Text(
            f"Final\n"
            f"Water: {water:.3f} gal\n"
            f"Additive: {additive:.5f} gal\n"
            f"Conc: {concentration:.5f}%",
            font_size=24,
        ).next_to(bar, RIGHT, buff=1)

        self.play(Transform(info, final_info))
        self.wait(2)
```
