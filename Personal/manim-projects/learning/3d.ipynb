```python
import manim as mn
from manim import *

config.media_width = "75%"
config.verbosity = "WARNING"

print(mn.__version__)
```

```python
%%manim -qm Axes3DExample

class Axes3DExample(ThreeDScene):
    def construct(self): 
        axes = ThreeDAxes()

        x_label = axes.get_x_axis_label(Tex("x"))
        y_label = axes.get_y_axis_label(Tex("y")).shift(UP * 1.8)

        # 3D variant of the Dot() object
        dot = Dot3D()

        # zoom out so we see the axes
        self.set_camera_orientation(zoom=0.5)

        self.play(FadeIn(axes), FadeIn(dot), FadeIn(x_label), FadeIn(y_label))

        self.wait(0.5)

        # animate the move of the camera to properly see the axes
        self.move_camera(phi=75 * DEGREES, theta=30 * DEGREES, zoom=1, run_time=1.5)

        # built-in updater which begins camera rotation
        self.begin_ambient_camera_rotation(rate=0.15)

        # one dot for each direction
        upDot = dot.copy().set_color(RED)
        rightDot = dot.copy().set_color(BLUE)
        outDot = dot.copy().set_color(GREEN)

        self.wait(1)

        self.play(
            upDot.animate.shift(UP),
            rightDot.animate.shift(RIGHT),
            outDot.animate.shift(OUT),
        )

        self.wait(2)
```

```python
%%manim -qm Rotation3DExample

class Rotation3DExample(ThreeDScene):
    def construct(self):
        cube = Cube(side_length=3, fill_opacity=0.8)

        self.begin_ambient_camera_rotation(rate=0.3)

        self.set_camera_orientation(phi=75 * DEGREES, theta=30 * DEGREES)

        self.play(Write(cube), run_time=2)

        self.wait(3)

        self.play(Unwrite(cube), run_time=2)
```

```python
%%manim -qm Basic3DExample

class Basic3DExample(ThreeDScene):
    def construct(self):
        cube = Cube(side_length=3, fill_opacity=0.5)

        self.set_camera_orientation(phi=75 * DEGREES, theta=30 * DEGREES)

        self.play(FadeIn(cube))

        for axis in [RIGHT, UP, OUT]:
            self.play(Rotate(cube, PI / 2, about_point=ORIGIN, axis=axis))

        self.play(FadeOut(cube))
```

```python
%%manim -qm BooleanOperations3DExample

config.verbosity = "CRITICAL"

class BooleanOperations3DExample(ThreeDScene):
    def construct(self):
        cube = Cube(side_length=3, fill_opacity=0.5)
        sphere = Sphere(radius=1.5, fill_opacity=0.5).shift(UP*0.75).shift(RIGHT*0.75).shift(OUT*0.75)

        # march camera from 0 to 180 phi and 0 to 360 theta
        self.set_camera_orientation(phi=0 * DEGREES, theta=0 * DEGREES)

        self.set_camera_orientation(phi=90 * DEGREES, theta=0 * DEGREES)


        self.play(FadeIn(cube), FadeIn(sphere))

        union = Union(cube, sphere, fill_opacity=0.5)
        intersection = Intersection(cube, sphere, fill_opacity=0.5)
        difference = Difference(cube, sphere, fill_opacity=0.5)
        exclusion = Exclusion(cube, sphere, fill_opacity=0.5)


        self.wait(1)


        self.wait(1)
```
